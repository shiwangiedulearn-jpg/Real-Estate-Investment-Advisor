```python
df[df["Area_sqft"]<=0]
df[df["Bedrooms"]<=0]
df[df["Bathrooms"]<=0]
df[df["Floors"]<=0]
df[df["Year_Built"]>2026]
df[df["Current_Price"]<=0]
```

```python
for col in df[["Area_sqft","Current_Price", "Price_After_5_Years"]]:
    Q1= df[col].quantile(0.25)
    Q3= df[col].quantile(0.75)
    IQR= Q3-Q1
    Lower_Bound= Q1-1.5*IQR
    Upper_Bound= Q3+1.5*IQR
    print(f"Column: {col}")
    print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
    print(f"Lower Bound: {Lower_Bound}, Upper Bound: {Upper_Bound}")
    print("Outlier: ", df[(df[col]<Lower_Bound) | (df[col]>Upper_Bound)])
```

```python
df["Price_per_sqft"]= df["Current_Price"]/df["Area_sqft"]
```
